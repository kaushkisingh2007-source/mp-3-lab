```python

import numpy as np
import matplotlib.pyplot as plt

# Input sequence
x = [1, 2, 3, 4]

N = len(x)

# DFT calculation
X = []

for k in range(N):
    s = 0
    for n in range(N):
        s = s + x[n] * np.exp(-2j * np.pi * k * n / N)
    X.append(s)

print("DFT of the sequence:")
for i in range(N):
    print(X[i])

# Magnitude of DFT
mag = np.abs(X)

# Plotting
plt.figure(figsize=(9, 4))

plt.subplot(1, 2, 1)
plt.stem(range(N), x)
plt.title("Input Sequence")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid()

plt.subplot(1, 2, 2)
plt.stem(range(N), mag)
plt.title("Magnitude of DFT")
plt.xlabel("Frequency Index")
plt.ylabel("|X(k)|")
plt.grid()

plt.tight_layout()
plt.show()
```
