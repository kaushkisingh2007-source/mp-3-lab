```python

import numpy as np
import matplotlib.pyplot as plt

# Define values of x and y
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)

# Create a grid
X, Y = np.meshgrid(x, y)

# Form complex number
z = X + 1j*Y

# Apply complex functions
w1 = z**2
w2 = z**3
w3 = np.exp(z)

# Create subplots
plt.figure(figsize=(10, 8))

# Original z-plane
plt.subplot(2, 2, 1)
plt.scatter(X, Y)
plt.title("z-plane")
plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.grid()

# Mapping of z square
plt.subplot(2, 2, 2)
plt.scatter(w1.real, w1.imag)
plt.title("w = z^2")
plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.grid()

# Mapping of z cube
plt.subplot(2, 2, 3)
plt.scatter(w2.real, w2.imag)
plt.title("w = z^3")
plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.grid()

# Mapping of exponential function
plt.subplot(2, 2, 4)
plt.scatter(w3.real, w3.imag)
plt.title("w = exp(z)")
plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.grid()

plt.tight_layout()
plt.show()
```
