```python
import numpy as np

# Limits of integration
a = 0
b = 1

# Function
def f(x):
    return x**2

# Points of Gaussian quadrature
x1 = (a+b)/2 - (b-a)/(2*np.sqrt(3))
x2 = (a+b)/2 + (b-a)/(2*np.sqrt(3))

# Calculate integral
I = (b-a)/2 * (f(x1)+f(x2))

print("Value of integral =", I)
```
