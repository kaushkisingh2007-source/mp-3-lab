```python
import numpy as np
import matplotlib.pyplot as plt

# Original point
x = 2
y = 1

# Angle of rotation
a = np.radians(45)

# Rotation matrix
R = np.array([[np.cos(a), -np.sin(a)],
              [np.sin(a), np.cos(a)]])

# Rotate the point
p = np.array([x, y])
q = np.dot(R, p)

# Plot original and rotated points
plt.plot([0, x], [0, y], 'bo-', label="Original")
plt.plot([0, q[0]], [0, q[1]], 'ro-', label="Rotated")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Rotation Matrix")
plt.legend()
plt.grid()
plt.axis('equal')
plt.show()

print("Rotated point =", q)
```
