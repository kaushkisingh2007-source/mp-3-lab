```python

import numpy as np
import matplotlib.pyplot as plt

# Number of points
N = 100

# Range of x
x = np.linspace(-5, 5, N)

# Given function
y = np.exp(-x**2)

# Calculate FFT
Y = np.fft.fft(y)

# Frequency values
freq = np.fft.fftfreq(N, x[1] - x[0])

# Shift zero frequency to centre
Y = np.fft.fftshift(Y)
freq = np.fft.fftshift(freq)

# Magnitude of FFT
mag = np.abs(Y)

print("FFT calculated successfully")
print("First five FFT values:")
print(Y[:5])

# Plotting
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Function exp(-x^2)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(freq, mag)
plt.title("FFT Magnitude Spectrum")
plt.xlabel("Frequency")
plt.ylabel("|F(k)|")
plt.grid()

plt.tight_layout()
plt.show()
```
