```python

import numpy as np
import matplotlib.pyplot as plt

# Enter the complex number
a = 1
b = 1
n = 3

# Calculate modulus and argument
r = np.sqrt(a*a + b*b)
theta = np.arctan2(b, a)

# Modulus of each root
R = r**(1/n)

print("Cartesian and Polar forms of roots")

for k in range(n):

    # Calculate angle of each root
    angle = (theta + 2*k*np.pi)/n

    # Calculate real and imaginary parts
    x = R*np.cos(angle)
    y = R*np.sin(angle)

    print("\nRoot", k+1)
    print("Cartesian form =", round(x, 4), "+",
          round(y, 4), "i")

    print("Polar form =", round(R, 4),
          "(cos", round(angle, 4),
          "+ i sin", round(angle, 4), ")")

    # Plot the root
    plt.plot(x, y, 'ro')
    plt.text(x, y, str(k+1))

plt.axhline(0)
plt.axvline(0)
plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.title("Cube Roots of a Complex Number")
plt.grid()
plt.show()
```
