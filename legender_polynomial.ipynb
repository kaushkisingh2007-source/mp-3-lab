```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return x**2

# Limits of integration
a = 0
b = 1
n = 3

# Gauss-Legendre points and weights
x, w = np.polynomial.legendre.leggauss(n)

# Change the limits
x1 = (b-a)*x/2 + (a+b)/2

# Calculate the integral
I = 0

for i in range(n):
    I = I + w[i]*f(x1[i])

I = I*(b-a)/2

print("Value of integral =", I)

# Plot the function
x = np.linspace(0, 1, 100)
y = f(x)

plt.plot(x, y, label="f(x) = x^2")
plt.fill_between(x, y, alpha=0.2)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gauss-Legendre Quadrature")
plt.grid()
plt.legend()
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return x**2

# Number of polynomials
n = 2

# Gauss-Legendre points and weights
x, w = np.polynomial.legendre.leggauss(5)

# Calculate coefficients
a = []

for k in range(n+1):

    # Legendre polynomial
    p = np.polynomial.legendre.legval(
        x, [0]*k + [1])

    # Calculate integral
    I = 0
    for i in range(len(x)):
        I = I + w[i]*f(x[i])*p[i]

    # Calculate coefficient
    c = ((2*k+1)/2)*I
    a.append(c)

    print("a", k, "=", round(c,4))

# Calculate expansion
x1 = np.linspace(-1, 1, 100)

f1 = a[0] + a[1]*x1 + a[2]*(3*x1**2-1)/2

# Original function
y = f(x1)

# Plot both functions
plt.plot(x1, y, label="Original Function")
plt.plot(x1, f1, '--', label="Legendre Expansion")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Legendre Polynomial Expansion")
plt.grid()
plt.legend()
plt.show()
```
