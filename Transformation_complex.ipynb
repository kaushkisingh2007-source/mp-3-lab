```python

import numpy as np
import matplotlib.pyplot as plt

# Original complex number
x = 2
y = 1

# Translation
a = 1
b = 2
x1 = x + a
y1 = y + b

# Scaling
s = 2
x2 = s*x
y2 = s*y

# Rotation by 90 degrees
theta = np.pi/2
x3 = x*np.cos(theta) - y*np.sin(theta)
y3 = x*np.sin(theta) + y*np.cos(theta)

# Reflection about real axis
x4 = x
y4 = -y

# Display coordinates
print("Original point:", x, y)
print("Translation:", x1, y1)
print("Scaling:", x2, y2)
print("Rotation:", round(x3,2), round(y3,2))
print("Reflection:", x4, y4)

# Plot all points
plt.plot(x, y, 'ro', label="Original")
plt.plot(x1, y1, 'bo', label="Translation")
plt.plot(x2, y2, 'go', label="Scaling")
plt.plot(x3, y3, 'mo', label="Rotation")
plt.plot(x4, y4, 'co', label="Reflection")

plt.xlabel("Real Part")
plt.ylabel("Imaginary Part")
plt.title("Complex Number Transformations")
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.legend()
plt.show()
```
