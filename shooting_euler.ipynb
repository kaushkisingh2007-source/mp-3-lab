```python
import numpy as np
import matplotlib.pyplot as plt

h = 0.01
n = 100

# Function for shooting method
def shoot(s):
    x = 0
    y = 0
    v = s

    X = [x]
    Y = [y]

    for i in range(n):
        y = y + h*v
        v = v - 2*h
        x = x + h

        X.append(x)
        Y.append(y)

    return y, X, Y


# Initial guesses for slope
s1 = 0
s2 = 2

# Secant method
for i in range(5):
    y1 = shoot(s1)[0]
    y2 = shoot(s2)[0]

    if abs(y2) < 0.000001:
        break

    s3 = s2 - y2*(s2-s1)/(y2-y1)

    s1 = s2
    s2 = s3

# Final solution
y, X, Y = shoot(s2)

print("Initial slope =", s2)
print("Final value =", y)

# Plot graph
plt.plot(X, Y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Shooting Method")
plt.grid()
plt.show()
```
