```python

import numpy as np
import matplotlib.pyplot as plt

# Number of points
n = 10

# Gauss-Laguerre method
x, w = np.polynomial.laguerre.laggauss(n)

I1 = 0

for i in range(n):
    I1 = I1 + w[i]*x[i]

# Analytical value
A1 = 1

print("Gauss-Laguerre Method")
print("Numerical value =", I1)
print("Analytical value =", A1)

# Gauss-Hermite method
x, w = np.polynomial.hermite.hermgauss(n)

I2 = 0

for i in range(n):
    I2 = I2 + w[i]

# Analytical value
A2 = np.sqrt(np.pi)

print("\nGauss-Hermite Method")
print("Numerical value =", I2)
print("Analytical value =", A2)

# Plot Gauss-Laguerre integrand
x1 = np.linspace(0, 8, 100)
y1 = x1*np.exp(-x1)

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(x1, y1)
plt.title("Gauss-Laguerre Method")
plt.xlabel("x")
plt.ylabel("x exp(-x)")
plt.grid()

# Plot Gauss-Hermite integrand
x2 = np.linspace(-4, 4, 100)
y2 = np.exp(-x2**2)

plt.subplot(1,2,2)
plt.plot(x2, y2)
plt.title("Gauss-Hermite Method")
plt.xlabel("x")
plt.ylabel("exp(-x^2)")
plt.grid()

plt.tight_layout()
plt.show()
```
